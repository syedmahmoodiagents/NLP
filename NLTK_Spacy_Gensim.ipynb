| ------------------ | ------------------------------------- | ------------------------------------------ |

Count Vectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
documents = [
    "The sky is blue.",
    "The sun is bright.",
    "The sun in the sky is bright.",
    "We can see the shining sun, the bright sun."
]
```

One Hot Encoding

```python
ohevector = CountVectorizer(binary=True)
```

```python
ohevector.fit_transform(documents)
```

```python
ohevector.get_feature_names_out()
```

```python
ohevector.fit_transform(documents).toarray()
```

Bag of Words

```python
# Create the vectorizer
vectorizer = CountVectorizer(binary=False) # by default it remains false
```

```python
# Fit and transform the documents
X = vectorizer.fit_transform(documents)
```

```python
# Show feature names (vocabulary)
print("Vocabulary:")
print(vectorizer.get_feature_names_out())
```

```python
# Show BoW representation
print("Count Vectors (sparse matrix):")
print(X.toarray())
```

TF-IDF

```python
# Create the vectorizer
tfidf_vectorizer = TfidfVectorizer()
```

```python
# Fit and transform the documents
X_tfidf = tfidf_vectorizer.fit_transform(documents)
```

```python
# Show TF-IDF feature names
print("TF-IDF Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())
```

```python
# Show TF-IDF values
print("TF-IDF Vectors:")
print(X_tfidf.toarray())
```

```python
from gensim.models import Word2Vec
```

```python
sentences = [['the', 'cat', 'sat', 'on', 'the', 'mat'],
            ['the', 'dog', 'barked', 'at', 'the', 'cat'],
            ['the', 'dog', 'sat', 'on', 'the', 'rug'],
            ['the', 'mat', 'was', 'sat', 'on', 'by', 'the', 'cat']]
```

```python
model = Word2Vec(sentences, vector_size=7, window=5, min_count=1)
```

```python
print(model.wv["cat"])
```

```python
model.wv.most_similar("cat")
```
