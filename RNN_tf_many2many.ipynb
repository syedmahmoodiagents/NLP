```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
# sentences with 11 unique words
sentences = [
    "i love deep learning",
    "i love machine learning",
    "deep learning is powerful",
    "machine learning is useful",
    "i enjoy learning ai",
    "i enjoy engineering"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
vocab_size = len(tokenizer.word_index) + 1 # extra 1 for padding
```

```python
vocab_size
```

```python
sequences = tokenizer.texts_to_sequences(sentences)
print(sequences)
```

```python
X = []
Y = []

for seq in sequences:
    X.append(seq[:-1])     # Input
    Y.append(seq[1:])      # Shifted Target

for a,b in zip(X,Y):
  print(a,b)
```

```python
max_len = max(len(x) for x in X)
print(max_len)
```

```python
X = pad_sequences(X, maxlen=max_len, padding='post')
Y = pad_sequences(Y, maxlen=max_len, padding='post')
```

```python
X
```

```python
print("Input Shape :", X.shape)
print("Target Shape:", Y.shape)
```

```python
# return_sequences=True
# the RNN layer outputs a sequence of hidden states,
# where each hidden state corresponds to an input in the sequence.
# This is typically used when you're stacking multiple RNN layers together,
# or when you need the output at each step
```

```python
model = Sequential([
    Embedding(vocab_size, 64),
    SimpleRNN(128, return_sequences=True),
    Dense(vocab_size, activation='softmax')
])
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X, Y, epochs=300)
```

```python
inpt = "i love deep"

seq = tokenizer.texts_to_sequences([inpt])[0]
seq = pad_sequences([seq], maxlen=max_len, padding='post')

print(seq)
```

```python
prediction = model.predict(seq, verbose=0)
```

```python
prediction[0]
```

```python
predicted_ids = np.argmax(prediction[0], axis=-1)
print(predicted_ids)
```

```python
predicted_words = [tokenizer.index_word.get(i, "-") for i in predicted_ids]
print(predicted_words)
```

```python
print("\nInput Words          :", inpt.split())
print("Predicted Next Words :", predicted_words)
```
