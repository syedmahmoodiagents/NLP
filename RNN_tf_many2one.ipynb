```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
sentences = [
    "i love deep learning",
    "i love machine learning",
    "deep learning is powerful",
    "machine learning is useful",
    "i enjoy learning ai"
]
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

word_index = tokenizer.word_index
vocab_size = len(word_index) + 1

print(word_index)
```

```python
sequences = tokenizer.texts_to_sequences(sentences)
print(sequences)
```

```python
X = []
y = []
for seq in sequences:
    for i in range(1, len(seq)):
        X.append(seq[:i])
        y.append(seq[i])
```

```python
for a,b in zip(X, y):
    print(a,b)
```

```python
X = pad_sequences(X, padding="pre")
X = np.array(X)
y = np.array(y)

print(X)
print(y)
```

```python
model = Sequential([
    Embedding(vocab_size, 64),
    SimpleRNN(128, return_sequences=False),
    Dense(vocab_size, activation='softmax')
])
```

```python
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
```

```python
# return_sequences=False
# the RNN layer will only output the hidden state from the last timestep of the input sequence
```

```python
model.fit(X, y, epochs=200)
```

```python
seq = tokenizer.texts_to_sequences(["i love"])[0]
seq = pad_sequences([seq], maxlen=X.shape[1], padding='pre')
```

```python
pred = model.predict(seq)
print(pred)
```

```python
np.argmax(pred)
```

```python
word = tokenizer.index_word[np.argmax(pred)]
```
